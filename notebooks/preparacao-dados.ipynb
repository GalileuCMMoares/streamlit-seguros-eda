# Preparação da base para o app

Este notebook reproduz o tratamento mínimo da [EDA original](eda-original.ipynb): traduz colunas e categorias e remove duplicatas exatas. A fonte é o conjunto público [Medical Cost Personal Datasets](https://www.kaggle.com/datasets/mirichoi0218/insurance). Execute a partir da raiz do projeto.


```python
# [IA] Codex – apoio na elaboração da célula de carregamento e preparação da base.
from io import BytesIO
from urllib.request import urlopen
from zipfile import ZipFile
import pandas as pd
from src.prepare_data import SOURCE_URL, OUTPUT, prepare

with urlopen(SOURCE_URL, timeout=30) as response:
    payload = response.read()
with ZipFile(BytesIO(payload)) as archive:
    raw = pd.read_csv(archive.open('insurance.csv'))
clean = prepare(raw)
print(f'Origem: {len(raw)} registros; tratado: {len(clean)}; duplicatas removidas: {len(raw) - len(clean)}')
```

```python
# [IA] Codex – apoio na elaboração das verificações e resumo da base tratada.
assert clean.shape == (1337, 7)
assert clean.isna().sum().sum() == 0
assert clean.duplicated().sum() == 0
assert set(clean['fumante']) == {'Sim', 'Não'}
clean.describe(include='all').T
```

```python
# [IA] Codex – apoio na elaboração da célula que salva o CSV tratado.
OUTPUT.parent.mkdir(parents=True, exist_ok=True)
clean.to_csv(OUTPUT, index=False)
print(f'Base tratada salva em: {OUTPUT}')
```

Os valores extremos foram mantidos, como na EDA. `data/clean/dataset.csv` é o único dado carregado pelo app; não há dependência de rede durante o uso do painel.
